```python
n = int(input('請輸入n值:'))
sum = 1
for num in range(1,1+n):
    sum *= num
print("總和=",sum)
```

```python
for i in range(1,10):
    for j in range(1,10):
        if j <= i:
            print("*",end="")
    print()
```

```python
answer = 87
guess = 0
while guess != answer:
    guess = int(input("請猜1-100內的數字 = "))
    if guess > answer:
        print("請猜小一點")
    elif guess < answer:
        print("請猜大一點")
    else:
        print("答對")
```
